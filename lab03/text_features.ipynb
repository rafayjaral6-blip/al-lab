# Laboratory 3: Text Features

Part A demonstrates Bag-of-Words feature extraction with `CountVectorizer`. The corpus contains 40 sample documents.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "The sun is shining today.",
    "The weather is good, the sun is great.",
    "A sunny day is a wonderful day.",
    "Clouds drift slowly across the blue sky.",
    "A gentle breeze cools the warm afternoon.",
    "Rain tapped softly against the window.",
    "Bright flowers opened beside the garden path.",
    "Birds sang from the branches of an old tree.",
    "The morning air felt crisp and fresh.",
    "Dark clouds gathered over the distant hills.",
    "A rainbow appeared after the summer shower.",
    "The lake reflected the golden evening light.",
    "Children played outside during the sunny break.",
    "Leaves rustled in the cool autumn wind.",
    "The forecast calls for clear skies tomorrow.",
    "Snow covered the rooftops through the night.",
    "Warm sunlight filled the quiet kitchen.",
    "Heavy rain flooded the narrow country road.",
    "The coastal fog lifted before noon.",
    "A pale moon rose above the sleeping town.",
    "Thunder rumbled beyond the mountain range.",
    "Fresh snow sparkled under the morning sun.",
    "The fields turned green after several rainy days.",
    "A cool breeze moved through the open window.",
    "Golden leaves covered the walking trail.",
    "The sunrise painted the clouds pink and orange.",
    "Small waves rolled gently onto the sandy shore.",
    "The garden soil stayed damp after the storm.",
    "A flock of geese crossed the cloudy sky.",
    "Summer heat warmed the stones beside the river.",
    "The winter evening arrived early and quietly.",
    "Soft clouds shaded the park in the afternoon.",
    "The first spring flowers appeared near the fence.",
    "A strong wind bent the grass in the meadow.",
    "The rain stopped and sunlight returned at noon.",
    "Morning dew glistened on the green leaves.",
    "A distant lighthouse shone through the sea mist.",
    "The dry path cracked beneath the hot summer sun.",
    "Cool air flowed down from the snowy peaks.",
    "Stars emerged as the clear evening sky darkened.",
]

print(f"Number of documents: {len(documents)}")
for index, document in enumerate(documents, start=1):
    print(f"Doc {index}: {document}")

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
feature_names = vectorizer.get_feature_names_out()
text_matrix = X.toarray()
print(f"\nFeature matrix shape: {text_matrix.shape}")
display(pd.DataFrame(text_matrix, columns=feature_names, index=[f"Doc {i + 1}" for i in range(len(documents))]))
```

```python
print("Lowercase enabled:", vectorizer.lowercase)
print("Default token pattern:", vectorizer.token_pattern)
print("Single-character 'a' included:", 'a' in feature_names)
print(f"Matrix sparsity: {(text_matrix == 0).mean():.1%}")
```
